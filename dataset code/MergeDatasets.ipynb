```python
import os
import shutil
from PIL import Image
import pandas as pd
from tqdm import tqdm
import country_converter as coco

# -----------------------------------------------------------
# CONFIG
# -----------------------------------------------------------
OUTPUT_DIR = "merged_dataset"
TARGET_SIZE = (640, 640)

GEOGUESSR_DIR = "geoguessr_dataset/compressed_dataset"
STREETVIEW_DIR = "streetviewByCountry/streetview_images"
ARCHIVE_DIR = "archive/dataset"
HF_ROOT = "huggingFaceDataset/datasets--deboradum--GeoGuessr-countries"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# Convert country name → ISO2 country code
cc = coco.CountryConverter()

def to_iso2(label):
    try:
        code = cc.convert(names=label, to="ISO2")
        if code == "not found":
            return None
        return code
    except:
        return None
```

```python
# -----------------------------------------------------------
# Helper: Save resized image
# -----------------------------------------------------------
def save_resized(img_path, out_path):
    try:
        img = Image.open(img_path).convert("RGB")
        img = img.resize(TARGET_SIZE)
        img.save(out_path, "JPEG", quality=95)
    except Exception as e:
        print(f"Error processing {img_path}: {e}")

def save_resized_and_delete(src_path, dst_path):
    try:
        img = Image.open(src_path).convert("RGB")
        img = img.resize(TARGET_SIZE)
        img.save(dst_path, "JPEG", quality=95)
        img.close()
        # Delete original file
        os.remove(src_path)
    except Exception as e:
        print(f"Error processing {src_path}: {e}")

# -----------------------------------------------------------
# Helper: Ensure label folder exists
# -----------------------------------------------------------
def ensure_label_folder(code):
    folder = os.path.join(OUTPUT_DIR, code)
    os.makedirs(folder, exist_ok=True)
    return folder
```

```python
# -----------------------------------------------------------
# 1. geoguessr_dataset/compressed_dataset
# -----------------------------------------------------------
print("\nProcessing Dataset 1 (geoguessr)…")
for country_name in os.listdir(GEOGUESSR_DIR):
    in_dir = os.path.join(GEOGUESSR_DIR, country_name)
    if not os.path.isdir(in_dir):
        continue

    code = to_iso2(country_name)
    if code is None:
        print(f"Skipping unmappable country: {country_name}")
        continue

    out_dir = ensure_label_folder(code)

    for img_name in tqdm(os.listdir(in_dir), desc=country_name):
        src = os.path.join(in_dir, img_name)
        dst = os.path.join(out_dir, f"geo_{img_name}.jpg")
        save_resized_and_delete(src, dst)

    # Remove empty folder
    try:
        os.rmdir(in_dir)
    except:
        pass
```

```python
# -----------------------------------------------------------
# 2. streetviewByCountry/streetview_images
# -----------------------------------------------------------
print("\nProcessing Dataset 2 (streetview)…")
for code in os.listdir(STREETVIEW_DIR):
    in_dir = os.path.join(STREETVIEW_DIR, code)
    if not os.path.isdir(in_dir):
        continue

    if len(code) != 2:
        print(f"Invalid country folder: {code}")
        continue

    out_dir = ensure_label_folder(code)

    for img_name in tqdm(os.listdir(in_dir), desc=code):
        src = os.path.join(in_dir, img_name)
        dst = os.path.join(out_dir, f"street_{img_name}.jpg")
        save_resized_and_delete(src, dst)

    try:
        os.rmdir(in_dir)
    except:
        pass
```

```python
# -----------------------------------------------------------
# 3. archive/dataset + CSV
# -----------------------------------------------------------
print("\nProcessing Dataset 3 (archive)…")

csv_path = os.path.join(ARCHIVE_DIR, "coords_with_countries.csv")
df = pd.read_csv(csv_path)

for _, row in tqdm(df.iterrows(), total=len(df)):
    img_id = str(row[0])
    country_code = str(row["country_code"]).upper()

    img_path = os.path.join(ARCHIVE_DIR, f"{img_id}.png")
    if not os.path.exists(img_path):
        continue

    out_dir = ensure_label_folder(country_code)
    dst = os.path.join(out_dir, f"arch_{img_id}.jpg")

    save_resized_and_delete(img_path, dst)

# (optional) delete CSV if you want
# os.remove(csv_path)
```

```python
# -----------------------------------------------------------
# 4. HuggingFace dataset — by default we DO NOT DELETE originals
# -----------------------------------------------------------
print("\nProcessing Dataset 4 (HuggingFace)…")

snapshot_root = os.path.join(HF_ROOT, "snapshots")
commit_dirs = [
    os.path.join(snapshot_root, d) 
    for d in os.listdir(snapshot_root)
    if os.path.isdir(os.path.join(snapshot_root, d))
]

if not commit_dirs:
    print("No snapshot found.")
else:
    snap = commit_dirs[0]

    for root, dirs, files in os.walk(snap):
        for f in files:
            if f.lower().endswith((".jpg", ".jpeg", ".png")):
                src = os.path.join(root, f)

                # Infer label
                parts = src.replace("\\", "/").split("/")
                maybe_label = parts[-2]

                code = maybe_label.upper() if len(maybe_label) == 2 else to_iso2(maybe_label)
                if code is None:
                    continue

                out_dir = ensure_label_folder(code)
                dst = os.path.join(out_dir, f"hf_{f}.jpg")

                # Save resized but DO NOT DELETE HuggingFace originals
                try:
                    img = Image.open(src).convert("RGB")
                    img = img.resize(TARGET_SIZE)
                    img.save(dst, "JPEG", quality=95)
                    img.close()
                except Exception as e:
                    print(f"Error HF image {src}: {e}")

print("\n🎉 FINISHED — Disk space optimized and dataset merged!")
```
