```python
import pandas as pd
import reverse_geocoder as rg

# Load your CSV (assumes two columns: latitude, longitude)
df = pd.read_csv("coords.csv", names=["latitude", "longitude"])

coords = list(zip(df["latitude"], df["longitude"]))

# Reverse geocode OFFLINE
results = rg.search(coords)  # returns list of dicts

# Extract country codes (e.g. "US", "FR")
country_codes = [res["cc"] for res in results]

# Add to dataframe
df["country_code"] = country_codes

# Optional: convert country codes -> country names
import pycountry
def get_country_name(code):
    try:
        return pycountry.countries.get(alpha_2=code).name
    except:
        return "Unknown"

df["country_name"] = df["country_code"].apply(get_country_name)

# Save CSV
df.to_csv("coords_with_countries.csv", index=False)

print("Done! Saved as coords_with_countries.csv")
```
