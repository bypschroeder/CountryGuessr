```python
import os
import requests

def download_with_resume(url, target_path):
    # check existing file
    existing = 0
    if os.path.exists(target_path):
        existing = os.path.getsize(target_path)

    headers = {}
    if existing > 0:
        headers["Range"] = f"bytes={existing}-"

    with requests.get(url, headers=headers, stream=True) as r:
        if r.status_code in (200, 206):
            mode = "ab" if existing else "wb"
            with open(target_path, mode) as f:
                for chunk in r.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
        else:
            raise RuntimeError(f"Download failed: {r.status_code}")

# usage
download_with_resume("https://scontent-fra3-2.xx.fbcdn.net/m1/v/t6/An8t-3LMOE6eyv7YZ5CD_ReDw-_DCGg2KLySdox-cFJUwtaKsQRjSRgz6gvDgaBpJnkfXoFHLGSTWnfmV5-ofbaqWwlduOhznfVmA0z5xBdLwcqKSXTdDLl1LjZo.md5?_nc_gid=o8D-PPkbm3bbMsJgk9xobA&_nc_oc=Adl2YbCR_VJ0kCf74ur1dmpYmPrTc2IDqI13YAqZK50zTcv11jzGduMcbxwewKvLEos&ccb=10-5&oh=00_AffctXPAD77T8SwFjBfWoUf3nkm0TFJupEnI-Fnsomf8jg&oe=6925B64C&_nc_sid=6de079", "msls_part1.zip")
```
