```python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("GOOGLE_API_KEY")

if API_KEY is None:
    raise ValueError("API_KEY not found in environment variables!")

print("API Key loaded successfully!")
```

```python
from geopy.geocoders import Nominatim

# TODO: Get good locations for test images
test_locations = [
    "Hof University of Applied Sciences, Hof, Germany",
    "Dijon, France",
    "York Minster, York, United Kingdom", 
    "Parma, Italy",
    "Plaza Mayor, Salamanca, Spain",
    "University of Coimbra, Coimbra, Portugal",
    "State Capitol, Madison, Wisconsin, USA",
    "University of Guelph, Guelph, Ontario, Canada",
    "Botanical Garden, Curitiba, Brazil",
    "Stellenbosch University, Stellenbosch, South Africa",
    "Lake Wendouree, Ballarat, Australia",
    "Matsumoto Castle, Matsumoto, Japan",
    "Gwangju, South Korea",
    "Yogyakarta, Indonesia",
    "Lenin Square, Novosibirsk, Russia",
    "Hagia Sophia, Istanbul, Turkey",
    "Bhopal Railway Station, Bhopal, India",
    "Luxor Temple, Luxor, Egypt",
    "Plaza 9 de Julio, Salta, Argentina"
]

locations = []

geolocator = Nominatim(user_agent="country_predictor", timeout=10)

for loc in test_locations:
    try:
        location = geolocator.geocode(loc)
        if location:
            locations.append((location.latitude, location.longitude, loc))
        else:
            print(f"No coordinates found for {loc}")
    except Exception  as e:
        print(f"Error for {loc}: {e}")

for lat, lon, name in locations:
    print(f"  {name}: {lat:.4f}, {lon:.4f}")
```

```python
import requests
import os
from time import sleep

BASE_URL = "https://maps.googleapis.com/maps/api/streetview"

output_dir = "streetview_images"
os.makedirs(output_dir, exist_ok=True)

for lat, lon, name in locations:
    safe_name = "".join(c for c in name if c.isalnum() or c in (' ', '-', '_')).rstrip()
    location_dir = os.path.join(output_dir, safe_name[:50])  # Limit length
    os.makedirs(location_dir, exist_ok=True)
    
    for heading in [0, 90, 180, 270]:
        params = {
            "size": "640x640",
            "location": f"{lat},{lon}",
            "heading": heading,
            "fov": 90,
            "key": API_KEY
        }
        response = requests.get(BASE_URL, params=params)
    
        if response.status_code == 200:
            filename = f"{lat:.3f}_{lon:.3f}_h{heading}.jpg"
            filepath = os.path.join(location_dir, filename)
            with open(filepath, 'wb') as f:
                f.write(response.content)
            print(f"Saved: {filename}")
        else:
            print(f"Error fetching image for {country} at {lat},{lon}, heading {heading}")
            
            sleep(0.1)
```
